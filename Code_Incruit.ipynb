```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import os

pd.set_option('display.max_colwidth', None)
```

```python
# 인크루트 데이터분석 검색
url = "https://search.incruit.com/list/search.asp?col=job&kw=%B5%A5%C0%CC%C5%CD%BA%D0%BC%AE"
# %B5%A5%C0%CC%C5%CD%BA%D0%BC%AE = 데이터분석
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# 공고 리스트 파싱
job_list = soup.select("ul.c_row")

data = []
for job in job_list:
    company_tag = job.select_one("a.cpname")
    company = company_tag.get_text(strip=True) if company_tag else ''

    title_tag = job.select_one("div.cell_mid a")
    title = title_tag.get_text(strip=True) if title_tag else ''

    # detail_tags = job.select("div.cl_md span")
    # job_field_tag = job.select_one("div.cl_btm span")

    # detail = [tag.get_text(strip=True) for tag in detail_tags]
    # if job_field_tag:
    #     detail.append(job_field_tag.get_text(strip=True))

    detail_tags = job.select("div.cl_md span")
    job_field_tags = job.select("div.cl_btm span")

    detail = [tag.get_text(strip=True) for tag in detail_tags + job_field_tags]


    link = title_tag.get("href") if title_tag else ""
    full_url = link if link else ""


    if company and title :
        data.append({
            "Site" : "인크루트",
            "Col_Company" : company,
            "Col_Recruit": title,
            "Col_detail": detail,
            "Col_url": full_url
        })

df = pd.DataFrame(data)
display(df)
```
